```python
import json

def load_json(file_path):
    with open(file_path, 'r') as file:
        data = json.load(file)
    return data
```

# Load VQA json

```python
official_train_vqa_filename = '<CT_RATE_DATASET_DIR>/labels/vqa_orig/train_vqa.json'
official_train_vqa = load_json(official_train_vqa_filename)
print(len(official_train_vqa))
```

```python
# Fiter out all question without images
official_train_vqa = [item for item in official_train_vqa if 'image' in item.keys()]
print(len(official_train_vqa))
```

# Create VQA internal train and internal val

```python
# Load internal train and val volumes list
with open('<CT_RATE_DATASET_DIR>/labels/internal_train_volumes.txt', 'r') as f:
    internal_train_volumes = f.read().splitlines()
print(len(internal_train_volumes))

with open('<CT_RATE_DATASET_DIR>/labels/internal_val_volumes.txt', 'r') as f:
    internal_val_volumes = f.read().splitlines()
print(len(internal_val_volumes))
```

```python
def filter_json_by_volumes(data, volume_list):
    filtered_data = [item for item in data if item['image'] in volume_list]
    return filtered_data
    
def filter_json_by_question_type(data, question_type):
    filtered_data = [item for item in data if item['id'].startswith(question_type)]
    return filtered_data
```

```python
internal_train_vqa = filter_json_by_volumes(official_train_vqa, internal_train_volumes)
```

```python
internal_val_vqa = filter_json_by_volumes(official_train_vqa, internal_val_volumes)
print(len(internal_train_vqa), len(internal_val_vqa))
```

```python
# Save internal VQA splits as json
with open('<CT_RATE_DATASET_DIR>/labels/vqa_orig/internal_train_vqa.json', 'w') as f:
    json.dump(internal_train_vqa, f)
with open('<CT_RATE_DATASET_DIR>/labels/vqa_orig/internal_val_vqa.json', 'w') as f:
    json.dump(internal_val_vqa, f)

print(len(load_json('internal_train_vqa.json')), len(load_json('internal_val_vqa.json')))
```

# Split the VQA files by question types

```python
report_generation_train = filter_json_by_question_type(internal_train_vqa, 'report_generation')
report_generation_valid = filter_json_by_question_type(internal_val_vqa, 'report_generation')
with open('report_generation_train.json', 'w') as f:
    json.dump(report_generation_train, f)
with open('report_generation_valid.json', 'w') as f:
    json.dump(report_generation_valid, f)
```

```python
short_answer_train = filter_json_by_question_type(internal_train_vqa, 'short_answer')
short_answer_valid = filter_json_by_question_type(internal_val_vqa, 'short_answer')
with open('short_answer_train.json', 'w') as f:
    json.dump(short_answer_train, f)
with open('short_answer_valid.json', 'w') as f:
    json.dump(short_answer_valid, f)
```

```python
mc_train = filter_json_by_question_type(internal_train_vqa, 'multiple_choice')
mc_valid = filter_json_by_question_type(internal_val_vqa, 'multiple_choice')
with open('multiple_choice_train.json', 'w') as f:
    json.dump(mc_train, f)
with open('multiple_choice_valid.json', 'w') as f:
    json.dump(mc_valid, f)
```

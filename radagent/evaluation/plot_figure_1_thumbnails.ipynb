```python
ct_scan = CT_RATE_ROOT / 'valid_fixed/dataset/valid_fixed/example_volume.nii.gz'
heart_segmentation = CT_RATE_PROCESSED_DIR / 'saved_segmentations/example_heart_segmentation.nii.gz'
aorta_segmentation = CT_RATE_PROCESSED_DIR / 'saved_segmentations/example_aorta_segmentation.nii.gz'
from tools.ct_utils import load_nifti
import numpy as np
from matplotlib.colors import ListedColormap
ct_data, ct_affine = load_nifti(ct_scan)
heart_data, heart_affine = load_nifti(heart_segmentation)
aorta_data, aorta_affine = load_nifti(aorta_segmentation)
print("CT scan shape:", ct_data.shape)
print("Heart segmentation shape:", heart_data.shape)
# select slice with biggest heart area
heart_area_per_slice = heart_data.sum(axis=(0, 1))
slice_with_biggest_heart = heart_area_per_slice.argmax()
print("Slice with biggest heart area:", slice_with_biggest_heart)
# display the slice with biggest heart area
plt.imshow(ct_data[:, :, slice_with_biggest_heart].T, cmap='gray', origin='lower')
# add overlay of heart segmentation in red
heart_mask = heart_data[:, :, slice_with_biggest_heart] > 0
masked_seg_map = np.ma.masked_where(heart_mask == 0, heart_mask)
plt.imshow(masked_seg_map.T, cmap=ListedColormap(['orange']), alpha=0.6, origin='lower', vmax=2)
# add overlay of aorta segmentation in blue
aorta_mask = aorta_data[:, :, slice_with_biggest_heart] > 0
masked_seg_map = np.ma.masked_where(aorta_mask == 0, aorta_mask)
plt.imshow(masked_seg_map.T, cmap=ListedColormap(['blue']), alpha=0.6, origin='lower')
plt.axis('off')
plt.savefig("figure_1_organ_segmentation.png", bbox_inches='tight', pad_inches=0, transparent=True, dpi=600)
plt.show()
```
